# Pydantic Basics: Creating and Using Models

Pydantic models are the foundation of data validation in Python. They use Python type annotations to define the structure and validate data at runtime. Here's detailed exploration of basic model creation with several examples

```python
from pydantic import BaseModel 
```

```python
class Person(BaseModel):
    name:str
    age:int
    city:str


person = Person(name="Bilal", age=30, city="Tetouan")
print(person)
```

```python
type(Person)
```

```python
from dataclasses import dataclass


# Here data vildation will not happen, because we are not inheriting from BaseModel
@dataclass
class Person:
    name:str
    age:int
    city:str

person = Person(name="Bilal", age=30, city=45)
print(person)
```

```python
type(person)
```

```python
class Person(BaseModel):
    name:str
    age:int
    city:str


person = Person(name="Bilal", age=30, city=45)
print(person)
```

## Model with Optional Fields

```python
from typing import Optional
class Employee(BaseModel):
    id:int
    name:str
    department:str
    salary:Optional[float]=None # salary is optional field
    is_active:Optional[bool]=True # is_active is optional field and default value is True
```

```python
employee = Employee(id=1, name="Bilal", department="IT")
print(employee)



```

```python
from pydantic import BaseModel
from typing import List
class Classroom(BaseModel):
    number_room:int
    students:List[str] # students is a list of strings
    capacity:int

classroom = Classroom(number_room=1, students=["Bilal", "Ali", "Ahmed"], capacity=30)
print(classroom)
```

## Model with nested values
Create complex structures with nested models


```python
from pydantic import BaseModel

class Address(BaseModel):
    street:str
    name:str
    zip_code: str

class Customer(BaseModel):
    customer_id:int
    name:str
    address:Address

customer = Customer(customer_id=1, name="Bilal", address=Address(street="123 Main St", name="Bilal", zip_code="12345"))
print(customer)





```

```python
customer = Customer(customer_id=1, name="Bilal", address=Address(street="123 Main St", name="Bilal", zip_code="12345"))
print(customer)
```

## Pydantic fields: customization and Constraints

The Field function in python enhances model fields beyond basic type hints by allowing you to specify validation rules, default values,
aliases, and more.

```python
from pydantic import BaseModel, Field

class Item(BaseModel):
    name:str=Field(min_length=3, max_length=50, description="The name of the item")
    price:float=Field(ge=0, le=1000, description="The price of the item")
    quantity:int=Field(default=1, ge=1, le=100, description="The quantity of the item")

item = Item(name="Laptop", price=1000, quantity=5)
print(item)
    
```

```python
from pydantic import BaseModel, Field
class User(BaseModel):
    username:str=Field(..., description="The username of the user") # three dots means it is required field
    age:int=Field(default=18, description="The age of the user, default=18")
    email:str=Field(default_factory=lambda: "default@example.com", description="The email of the user, default=default@example.com") # default_factory is a function that returns a default value

# difference between default and default_factory
# default is a static value
# default_factory is a function that returns a default value
```

```python
# Examples
user1 = User(username="Bilal")
print(user1)

user2 = User(username="Bilal", age=20, email="bilal@example.com")
print(user2)
```
