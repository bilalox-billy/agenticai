## Complete Example: From Chunking to Embedding


```python
from langchain_community.document_loaders import TextLoader

loader=TextLoader('examples/speech.txt')
docs=loader.load()
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
final_documents=text_splitter.split_documents(docs)
final_documents
```

```python
## Vector Embedding + Vector Store

from langchain_community.vectorstores import Chroma
# from_documents is a class method that takes a list of documents and an embedding model and outputs a Chroma database with
db = Chroma.from_documents(documents=final_documents, embedding=embedding_1024) 
```

```python
### Retrieve the results from query vectorstore db
query="It will be all the easier for us to conduct ourselves as belligerents"
retrieved_results=db.similarity_search(query)
print(retrieved_results)
```

## Ollama Embedding
Ollama supports embeddings models, making it possible to build retrieval augmented generatioon(RAG) applcation that combine text
prompts with existing documents or other data
Link: https://ollama.com/search?c=embedding

```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embedding = (
    OllamaEmbeddings(model="gemma:2b") # by default it uses the llama2 model
)
```

```python
embedding
```

```python
r1= embedding.embed_documents(
    [
        "Alpha is the first letter of Greek alphabet",
        "Beta is the second letter of Greek alphabet",
    ]
)
```

```python
len(r1[0])
```

```python
result = embedding.embed_query("What is the second letter of Greek alphabet?")
```

```python
len(result)
```
