```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['OPENAI_API_KEY']=os.getenv("OPEN_API_KEY")
## Langsmith Tracking
os.environ["LANGCHAIN_API_KEY"]=os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"]=os.getenv("LANGSMITH_TRACING")
os.environ["LANGCHAIN_PROJECT"]=os.getenv("LANGCHAIN_PROJECT")
os.environ["LANGCHAIN_ENDPOINT"]=os.getenv("LANGSMITH_ENDPOINT")
```

```python
from langchain_openai import ChatOpenAI
llm=ChatOpenAI(model="gpt-4o")
print(llm)
```

```python
## Input and get response from LLM

result = llm.invoke("What is generative AI")
```

```python
print(result)
```

```python
### ChatTemplate
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert AI Engineer. Provide me answers based on the question"),
    ("user", f"{input}")
])

prompt
```

```python
chain = prompt | llm
response = chain.invoke({"input": "Can you tell me about the latest trends in AI?"})
response
```

```python
## String output Parser: Get a message from LLM and convert it into a string
from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
chain = prompt | llm | output_parser
response=chain.invoke({"input": "Can you tell me about the latest trends in AI?"})
response
```
