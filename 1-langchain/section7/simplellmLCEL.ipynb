```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["LANGCHAIN_API_KEY"]=os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"]="true"
os.environ["LANGCHAIN_PROJECT"]=os.getenv("LANGCHAIN_PROJECT")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
model= ChatGroq(
    model="llama-3.1-8b-instant",
    groq_api_key=os.getenv("GROQ_API_KEY")
)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

messages= [
    SystemMessage(content="Transalte the following from English to French"),
    HumanMessage(content="I love programming in Python")
]

response= model.invoke(messages)


```

```python
from langchain_core.output_parsers import StrOutputParser
parser = StrOutputParser()

parser.invoke(response)
```

```python
### Using LCEL, we can chain the component
chain=model|parser
chain.invoke(messages)
```

```python
### Prompt Template
from langchain_core.prompts import ChatPromptTemplate

generic_template= "Translate the following into  {language}: "

prompt= ChatPromptTemplate.from_messages([
    ("system", generic_template),
    ("user", "{text}")
])

result = prompt.invoke({"text":"I love programming in Python", "language":"French"})

```

```python
result.to_messages()
```

```python
chain= prompt|model|parser
chain.invoke({"text":"I love programming in Python", "language":"French"})
```
